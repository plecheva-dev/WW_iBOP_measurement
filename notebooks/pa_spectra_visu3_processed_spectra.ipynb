```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../2_preprocessed_data/pa_spectra_preprocessed.csv")

wav = df.columns[1:].astype("float")

sample_list = df["sample"].unique()

short_sample_list = sorted(list(set([s[:3] for s in sample_list])))

fig, ax = plt.subplots(ncols=3, nrows=8, figsize=(12, 32), sharey=True)

for (i, sample) in enumerate(short_sample_list):

    sp_raw = df[df["sample"]==f"{sample}_raw"].values[0][1:]
    sp_15u = df[df["sample"]==f"{sample}_15u"].values[0][1:]
    sp_05u = df[df["sample"]==f"{sample}_05u"].values[0][1:]

    sp_calc = sp_15u+sp_05u

    ii = i//3
    jj = i%3

    ax[ii, jj].plot(wav, sp_raw, label="raw")
    ax[ii, jj].plot(wav, sp_15u, label="sp_15u")
    ax[ii, jj].plot(wav, sp_05u, label="sp_05u")
    ax[ii, jj].plot(wav, sp_calc, linestyle="--", label="15u+05u")
    ax[ii, jj].set_title(sample)
    

plt.tight_layout()
plt.show()



```
