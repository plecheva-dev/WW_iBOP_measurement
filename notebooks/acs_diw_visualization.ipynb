```python
import warnings
warnings.filterwarnings("ignore")

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import os

a_df = pd.read_csv("../data/processed/a_df.csv")
c_df = pd.read_csv("../data/processed/c_df.csv")
```

```python
a_df_diw = a_df[a_df["sample_type"] == "diw"]
a_df_diw_mean = a_df_diw.iloc[:, 5:89].mean()
a_df_diw_std = a_df_diw.iloc[:, 5:89].std()
c_df_diw = c_df[c_df["sample_type"] == "diw"]
c_df_diw_mean = c_df_diw.iloc[:, 5:89].mean()
c_df_diw_std = c_df_diw.iloc[:, 5:89].std()

wav = a_df_diw.columns[5:89].astype(float)

fig, ax = plt.subplots(ncols = 2, figsize=(8, 4), sharey=True)

ax[0].plot(wav, a_df_diw_mean, label='ACS-A DIW Mean')
ax[0].fill_between(wav, a_df_diw_mean - a_df_diw_std, a_df_diw_mean + a_df_diw_std, alpha=0.2, label='ACS-A DIW Std Dev')
ax[0].set_title('ACS-A DIW Spectra')
ax[0].set_xlabel('Wavelength (nm)')
ax[0].set_ylabel('Absorption Coefficient (1/m)')    
ax[0].legend()

ax[1].plot(wav, c_df_diw_mean, label='ACS-C DIW Mean')
ax[1].fill_between(wav, c_df_diw_mean - c_df_diw_std, c_df_diw_mean + c_df_diw_std, alpha=0.2, label='ACS-C DIW Std Dev')
ax[1].set_title('ACS-C DIW Spectra')
ax[1].set_xlabel('Wavelength (nm)')
ax[1].legend()

plt.ylim(-0.2, 1)
plt.tight_layout()
plt.show()
```
