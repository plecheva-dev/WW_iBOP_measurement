```python
### loading data

import pandas as pd
import matplotlib.pyplot as plt

a_df = pd.read_csv("../2_preprocessed_data/acs_a_extracted_data.csv", index_col=0)
c_df = pd.read_csv("../2_preprocessed_data/acs_c_extracted_data.csv", index_col=0)
```

```python
for s in a_df["sample"].unique(): 
    # get and plot data, saving the plot to "acs_triplicates_plots/"
    fig, ax = plt.subplots(ncols=2, figsize=(12,5))
    a_df_s = a_df[a_df["sample"]==s]
    c_df_s = c_df[c_df["sample"]==s]
    for i, row in a_df_s.iterrows():
        ax[0].plot(a_df_s.columns[2:].astype(float), row[2:], label=f'Run {row["run_nr"]}')
    for i, row in c_df_s.iterrows():
        ax[1].plot(c_df_s.columns[2:].astype(float), row[2:], label=f'Run {row["run_nr"]}')
    ax[0].set_title(f'Absorption Coefficient - Sample {s}')
    ax[0].set_xlabel('Wavelength (nm)')
    ax[0].set_ylabel('a (m$^{-1}$)')
    ax[0].legend()
    ax[1].set_title(f'Attenuation Coefficient - Sample {s}')
    ax[1].set_xlabel('Wavelength (nm)')
    ax[1].set_ylabel('c (m$^{-1}$)')
    ax[1].legend()
    plt.tight_layout()
    plt.savefig(f"acs_triplicates_plots/acs_triplicates_sample_{s}.png")
    plt.close()
```

```python
### loading data

import pandas as pd
import matplotlib.pyplot as plt

a_df = pd.read_csv("../2_preprocessed_data/acs_a_extracted_data_advanced.csv", index_col=0)
c_df = pd.read_csv("../2_preprocessed_data/acs_c_extracted_data_advanced.csv", index_col=0)

a_df_combined = pd.read_csv("../2_preprocessed_data/acs_a_extracted_data_combined_by_sample.csv", index_col=0)
c_df_combined = pd.read_csv("../2_preprocessed_data/acs_c_extracted_data_combined_by_sample.csv", index_col=0)

wv = a_df.columns[3:87].astype(float)

for s in a_df["sample"].unique(): 
    fig, ax = plt.subplots(ncols=2, figsize=(12,5))

    a_df_s = a_df[a_df["sample"]==s]
    c_df_s = c_df[c_df["sample"]==s]
    for i, row in a_df_s.iterrows():
        ax[0].plot(wv, row[3:87], label=f'Run {row["run_nr"]}, {row["n_valid_spectra"]} samples')
        lower_bound = row[3:87].values - row[87:].values
        upper_bound = row[3:87].values + row[87:].values
        ax[0].fill_between(
            wv, 
            lower_bound.astype("float"), 
            upper_bound.astype("float"), 
            alpha=0.2)
    for i, row in c_df_s.iterrows():
        ax[1].plot(wv, row[3:87], label=f'Run {row["run_nr"]}, {row["n_valid_spectra"]} samples')
        lower_bound = row[3:87].values - row[87:].values
        upper_bound = row[3:87].values + row[87:].values    
        ax[1].fill_between(
            wv, 
            lower_bound.astype("float"), 
            upper_bound.astype("float"), 
            alpha=0.2)
    
    s_simplified = s[-7:]

    if s_simplified in a_df_combined["sample_id"].values:
        for i, row in a_df_combined[a_df_combined["sample_id"]==s_simplified].iterrows():
            ax[0].plot(wv, row[2:86], color='black', linewidth=2, label=f'Combined Sample {s}, {row["n_valid_spectra"]} samples')
            lower_bound = row[2:86].values - row[86:].values
            upper_bound = row[2:86].values + row[86:].values
            ax[0].fill_between(
                wv, 
                lower_bound.astype("float"), 
                upper_bound.astype("float"), 
                color='black', alpha=0.2)
            
    if s_simplified in c_df_combined["sample_id"].values:
        for i, row in c_df_combined[c_df_combined["sample_id"]==s_simplified].iterrows():
            ax[1].plot(wv, row[2:86], color='black', linewidth=2, label=f'Combined Sample {s}, {row["n_valid_spectra"]} samples')
            lower_bound = row[2:86].values - row[86:].values
            upper_bound = row[2:86].values + row[86:].values    
            ax[1].fill_between(
                wv, 
                lower_bound.astype("float"), 
                upper_bound.astype("float"), 
                color='black', alpha=0.2)

    ax[0].set_title(f'Absorption Coefficient - Sample {s}')
    ax[0].set_xlabel('Wavelength (nm)')
    ax[0].set_ylabel('a (m$^{-1}$)')
    ax[0].legend()
    ax[1].set_title(f'Attenuation Coefficient - Sample {s}')
    ax[1].set_xlabel('Wavelength (nm)')
    ax[1].set_ylabel('c (m$^{-1}$)')
    ax[1].legend()
    plt.tight_layout()
    plt.savefig(f"acs_triplicates_plots_advanced/acs_triplicates_sample_{s}.png")
    plt.close()
```
